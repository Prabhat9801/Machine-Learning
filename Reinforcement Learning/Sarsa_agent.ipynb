```python
import numpy as np
import gymnasium as gym
```

```python
cliffEnv = gym.make("CliffWalking-v1")
```

```python
q_table = np.zeros(shape=(48,4))
```

```python
def policy(state,explore = 0.0):
  action = int(np.argmax(q_table[state]))
  if np.random.random()<=explore:
    action = np.random.randint(low=0,high=4)
  return action
```

```python
EPSILON = 0.1
ALPHA = 0.1
GAMMA = 0.9
NUM_EPISODES = 500
```

```python
for episode in range(NUM_EPISODES):
      done = False
      total_reward = 0
      episode_length = 0
      state, info = cliffEnv.reset()
      action = policy(state,EPSILON)
      while not done:
         next_state,reward,done,_, info = cliffEnv.step(action)
         next_action = policy(next_state,EPSILON)
         q_table[state][action] += ALPHA*(reward+GAMMA*q_table[next_state][next_action]-q_table[state][action])
         state = next_state
         action = next_action
         total_reward += reward
         episode_length += 1

      print("Episode is :",episode, "Episode_length:", episode_length, "Total Reward:", total_reward)
cliffEnv.close()
```

```python
import pickle as pkl
pkl.dump(q_table, open("q_table.pkl","wb"))
print("training complete and q_table is saved")
```

```python
import pickle
import numpy as np
import gymnasium as gym

# Load the saved Q-table
with open("q_table.pkl", "rb") as f:
    q_table = pickle.load(f)
```

```python
env = gym.make("CliffWalking-v1", render_mode="ansi")
state, info = env.reset()
done = False
```

```python
while not done:
    # Render the environment
    output = env.render()
    print(output)

    # Choose the best action from Q-table
    action = np.argmax(q_table[state])

    print(f"State {state} --> Action: {['Up','Right','Down','Left'][action]}")

    # Step in the environment
    state, reward, terminated, truncated, info = env.step(action)
    done = terminated or truncated

    print(f"New State: {state}, Reward: {reward}, Done: {done}")

env.close()
```
