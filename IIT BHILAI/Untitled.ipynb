```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import openpyxl
df = pd.read_csv('StudentMarksDataset.csv')
df
df = df.sort_values(by="Std_Name").reset_index(drop=True)
df.to_excel("Sorted_Student_Data.xlsx", index=False)
```

```python
df.info()
```

```python
df.columns = [
    "Timestamp",
    "College_Name",
    "Age",
    "Hometown_District",
    "Pincode",
    "Distance_Kms",
    "Travel_Time_Hrs",
    "JEE_Appeared",
    "Math_Comfort",
    "Python_Comfort",
    "Career_Goal",
    "Workshop_Expectations",
    "DS_Applications"
]
```

```python
import seaborn as sns
df["Distance_Kms"] = pd.to_numeric(df["Distance_Kms"], errors='coerce')

# Drop NaNs for clean plotting
distance_data = df["Distance_Kms"].dropna()

# Plot the box plot
plt.figure(figsize=(8, 5))
sns.boxplot(x=distance_data)
plt.title("Box Plot of Distance from IIT Bhilai")
plt.xlabel("Distance (Kms)")
plt.grid(True)
plt.show()
```

```python

df["Distance_Kms"] = pd.to_numeric(df["Distance_Kms"], errors='coerce')
distance_data = df["Distance_Kms"].dropna()

# Plot the distribution
plt.figure(figsize=(8, 5))
sns.histplot(distance_data, kde=True, bins=20, color='skyblue')
plt.title("Normal Distribution of Distance from IIT Bhilai")
plt.xlabel("Distance (Kms)")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```
